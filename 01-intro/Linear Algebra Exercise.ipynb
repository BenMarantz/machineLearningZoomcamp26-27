```python
# Import NumPy
import numpy as np
```

```python
# Matrix-Matrix Multiplication Function
def matrixMult(U, V):
    assert U.shape[1] == V.shape[0]
    result = np.zeros([U.shape[0], V.shape[1]])
    for i in range(U.shape[0]):
        for j in range(V.shape[1]):
            result[i,j] = sum(U[i,:] * V[:,j])
    return result
```

```python
# Example 1
U = np.array([
    [2, 4, 5, 6],
    [1, 2, 1, 2],
    [3, 1, 2, 1]
])

V = np.array([
    [1, 1, 2],
    [0, 0.5, 1],
    [0, 2, 1],
    [2, 1, 0]
])

matrixMult(U, V)
```

```python
# Check result with Numpy's dot function
np.dot(U, V)
```

```python
# Check result with Numpy's @ operator
U @ V
```

```python
# Check result with Numpy's matmul function
np.matmul(U, V)
```
