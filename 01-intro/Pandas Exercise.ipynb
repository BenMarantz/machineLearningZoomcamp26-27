```python
# Import NumPy and Pandas
import numpy as np
import pandas as pd
```

```python
# Create the data and inspect it.
data = [
    ['Nissan', 'Stanza', 1991, 138, 4, 'MANUAL', 'sedan', 2000],
    ['Hyundai', 'Sonata', 2017, None, 4, 'AUTOMATIC', 'Sedan', 27150],
    ['Lotus', 'Elise', 2010, 218, 4, 'MANUAL', 'convertible', 54990],
    ['GMC', 'Acadia',  2017, 194, 4, 'AUTOMATIC', '4dr SUV', 34450],
    ['Nissan', 'Frontier', 2017, 261, 6, 'MANUAL', 'Pickup', 32340],
]

columns = [
    'Make', 'Model', 'Year', 'Engine HP', 'Engine Cylinders',
    'Transmission Type', 'Vehicle_Style', 'MSRP'
]

df = pd.DataFrame(data, columns = columns)
df
```

```python
# Ways to get column. Say we want Make column
df.Make
```

```python
df['Make']
```

```python
# Ways to get row. Say we want first row
df.loc[0]
```

```python
# loc references the data frame's index attribute. The default is the number index (0 to n-1, where n is # of rows)
df.index
```

```python
# iloc always references the number index (0 to n-1, where n is # of rows), regardless of the actual index attribute
df.iloc[0]
```

```python
# Filtering
# Let's say we want rows were 'Year' is 2015 or later
df[df['Year'] >= 2015]
```

```python
# Say we want rows where there are no missing values.
df.dropna()
```

```python
# Alternatively, get rows with at least one missing value
df[df.isnull().any(axis=1)]
```

```python
# Say we want cars that are Sonatas made in 2015 or later
# When doing multiple comparisons, put each comparison in parentheses.
df[
    (df['Year'] >= 2015) & (df['Model'] == "Sonata")
]
```

```python
# Say we want cars that are either Nissan or Automatic
df[
    (df['Make'] == "Nissan") | (df['Transmission Type'] == "AUTOMATIC")
]
```

```python
# String Operations
# It's important to make sure column text is consistent
# Let's make it so 'Vehicle_Style' is all lowercase and has underscores in place of spaces
df['Vehicle_Style'] = df['Vehicle_Style'].str.replace(' ', '_').str.lower()
df['Vehicle_Style']
```

```python
# Summarize Data
# describe() function produces summary statistics of all numerical columns. 
df.describe()
```

```python
# Can still use the summary functions from NumPy, too
df.MSRP.mean()
```

```python
# Use nunique to get the number of categories in each categorical column
df.nunique()
```

```python
# Get number of missing values per column
df.isnull().sum()
```

```python
# Suppose we want the average MSRP of the cars grouped by transmission type. Use the groupby function.
df.groupby('Transmission Type').MSRP.mean()
```

```python
# Suppose we want average Engine HP grouped by number of Engine Cylinders
df.groupby('Engine Cylinders')['Engine HP'].mean()
```

```python
# To convert a data frame column to a NumPy array, use values function
df.MSRP.values
```

```python
# To convert a data frame to a dictionary, use to_dict(orient='records')
df.to_dict(orient='records')
```
